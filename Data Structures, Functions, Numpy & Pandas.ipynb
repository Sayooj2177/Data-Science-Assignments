### Question 1
Create a list of 5 integers and perform list operations.

```python
# Create a list of 5 integers
numbers = [5, 2, 8, 2, 1]

print("Original list:", numbers)

# append
numbers.append(10)
print("After append:", numbers)

# extend
numbers.extend([7, 9])
print("After extend:", numbers)

# insert
numbers.insert(2, 6)
print("After insert:", numbers)

# remove
numbers.remove(2)
print("After remove:", numbers)

# pop
removed_value = numbers.pop()
print("Popped value:", removed_value)
print("After pop:", numbers)

# index
print("Index of 8:", numbers.index(8))

# count
print("Count of 2:", numbers.count(2))

# sort
numbers.sort()
print("After sort:", numbers)

# reverse
numbers.reverse()
print("After reverse:", numbers)

# clear
numbers.clear()
print("After clear:", numbers)
```

## Section C: Exploring Pandas

### Question 11
Create a small DataFrame manually with 10 rows and columns: 'Name', 'Age', 'City', and 'Salary'.

Perform the following operations:
- Use .info() and .describe()
- Select 'Name' and 'City' columns
- Drop 'City' column
- Fill missing values in 'Salary' with the mean
- Remove duplicate rows

```python
# Create a DataFrame manually with 10 rows
data = {
    "Name": [
        "John", "Alice", "David", "Sarah", "Mike",
        "Emma", "Robert", "Lisa", "Daniel", "John"
    ],
    "Age": [
        22, 25, 30, 28, 35,
        24, 32, 27, 29, 22
    ],
    "City": [
        "Kochi", "Mumbai", "Delhi", "Chennai", "Pune",
        "Kochi", "Bangalore", "Mumbai", "Delhi", "Kochi"
    ],
    "Salary": [
        30000, 45000, 50000, None, 55000,
        40000, 60000, None, 48000, 30000
    ]
}

df_pandas = pd.DataFrame(data)

print("Original DataFrame:")
print(df_pandas)
```

```python
# Display DataFrame information
print("DataFrame Information:")
df_pandas.info()

# Display descriptive statistics
print("\nDescriptive Statistics:")
print(df_pandas.describe())
```

```python
# Select Name and City columns
selected_columns = df_pandas[["Name", "City"]]

print("Selected Name and City columns:")
print(selected_columns)
```

```python
# Drop the City column
df_without_city = df_pandas.drop("City", axis=1)

print("DataFrame after dropping City:")
print(df_without_city)
```

```python
# Calculate the mean salary
salary_mean = df_pandas["Salary"].mean()

# Fill missing Salary values with the mean
df_pandas["Salary"] = df_pandas["Salary"].fillna(salary_mean)

print("Mean Salary:", salary_mean)
print("\nDataFrame after filling missing Salary values:")
print(df_pandas)
```

```python
# Remove duplicate rows
df_no_duplicates = df_pandas.drop_duplicates()

print("DataFrame after removing duplicate rows:")
print(df_no_duplicates)
```
