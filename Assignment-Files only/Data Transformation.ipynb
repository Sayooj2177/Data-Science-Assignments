```python
df_encoded = df.copy()

df_encoded.head()
```

```python
from sklearn.preprocessing import OneHotEncoder

encoder = OneHotEncoder(sparse_output=False, drop=None)

sex_encoded = encoder.fit_transform(df_encoded[["sex"]])

sex_encoded_df = pd.DataFrame(
    sex_encoded,
    columns=encoder.get_feature_names_out(["sex"]),
    index=df_encoded.index
)

sex_encoded_df.head()
```

```python
df_encoded = pd.concat([df_encoded, sex_encoded_df], axis=1)

df_encoded.head()
```

```python
from sklearn.preprocessing import LabelEncoder

label_encoder = LabelEncoder()

label_columns = [
    "workclass",
    "education",
    "marital_status",
    "occupation",
    "relationship",
    "native_country"
]

for col in label_columns:
    df_encoded[col] = label_encoder.fit_transform(df_encoded[col])

df_encoded.head()
```

```python
df_encoded.dtypes
```

```python
from sklearn.preprocessing import LabelEncoder

# Label encode race
race_encoder = LabelEncoder()
df_encoded["race"] = race_encoder.fit_transform(df_encoded["race"])

# Encode income target
income_encoder = LabelEncoder()
df_encoded["income"] = income_encoder.fit_transform(df_encoded["income"])

# Remove original sex column because One-Hot encoded columns were created
df_encoded = df_encoded.drop(columns=["sex"])

df_encoded.head()
```

```python
df_encoded.dtypes
```

```python
X = df_encoded.drop(columns=["income"])
y = df_encoded["income"]

X.head()
```

```python
from sklearn.preprocessing import StandardScaler

standard_scaler = StandardScaler()

X_standard_scaled = standard_scaler.fit_transform(X)

X_standard_scaled = pd.DataFrame(
    X_standard_scaled,
    columns=X.columns,
    index=X.index
)

X_standard_scaled.head()
```

```python
from sklearn.preprocessing import MinMaxScaler

minmax_scaler = MinMaxScaler()

X_minmax_scaled = minmax_scaler.fit_transform(X)

X_minmax_scaled = pd.DataFrame(
    X_minmax_scaled,
    columns=X.columns,
    index=X.index
)

X_minmax_scaled.head()
```

```python
df_features = df.copy()

# Feature 1: Age group
df_features["age_group"] = pd.cut(
    df_features["age"],
    bins=[0, 25, 40, 60, 100],
    labels=["Young", "Adult", "Middle_Aged", "Senior"]
)

# Feature 2: Capital net
df_features["capital_net"] = (
    df_features["capital_gain"] - df_features["capital_loss"]
)

df_features[["age", "age_group", "capital_gain", "capital_loss", "capital_net"]].head()
```

```python
df_features["capital_gain"].skew()
```

```python
import numpy as np

df_features["capital_gain_log"] = np.log1p(df_features["capital_gain"])

df_features[["capital_gain", "capital_gain_log"]].head()
```
