```python
#1
number = int(input("Enter a number: "))

if number <= 1:
    print("Not a prime number")
else:
    is_prime = True

    for i in range(2, number):
        if number % i == 0:
            is_prime = False
            break

    if is_prime:
        print(number, "is a prime number")
    else:
        print(number, "is not a prime number")
```

```python
#2
import random

number1 = random.randint(1, 10)
number2 = random.randint(1, 10)

correct_answer = number1 * number2

print("What is", number1, "x", number2, "?")

user_answer = int(input("Enter your answer: "))

if user_answer == correct_answer:
    print("Correct answer!")
else:
    print("Wrong answer.")
    print("The correct answer is:", correct_answer)
```

```python
#3
# Choice: Printing squares of even numbers from 100 to 200

for number in range(100, 201):
    if number % 2 == 0:
        square = number ** 2
        print(number, "=", square)
```

```python
#4
# Hospital Billing System

room_charge_per_day = 2000
number_of_days = 4

doctor_fee = 1500

cost_per_lab_test = 300
number_of_tests = 3

medicine_charges = 2400

# Calculate individual charges
room_charges = room_charge_per_day * number_of_days
lab_charges = cost_per_lab_test * number_of_tests

# Calculate total before discount
total_bill = room_charges + doctor_fee + lab_charges + medicine_charges

# Calculate discount
if total_bill > 10000:
    discount = total_bill * 0.10
else:
    discount = 0

# Calculate final bill
final_bill = total_bill - discount

print("Hospital Billing System")
print("----------------------")
print("Room Charges: ₹", room_charges)
print("Doctor Fee: ₹", doctor_fee)
print("Lab Charges: ₹", lab_charges)
print("Medicine Charges: ₹", medicine_charges)
print("Total Bill: ₹", total_bill)
print("Discount: ₹", discount)
print("Final Bill: ₹", final_bill)
```
